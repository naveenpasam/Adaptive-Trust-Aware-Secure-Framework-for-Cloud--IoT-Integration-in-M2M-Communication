```python
import matplotlib.pyplot as plt
import numpy as np

metrics = ['Trust Accuracy', 'Latency', 'Detection Rate']

proposed = [96, 150, 95]
existing = [80, 240, 77]

x = np.arange(len(metrics))
width = 0.35

plt.figure()
plt.bar(x - width/2, proposed, width, label='Proposed Model')
plt.bar(x + width/2, existing, width, label='Existing Model')

plt.xlabel('Performance Metrics')
plt.ylabel('Values')
plt.title('Overall Performance Comparison')
plt.xticks(x, metrics)
plt.legend()
plt.grid(axis='y')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define data FIRST (this was missing)
devices = np.arange(10, 110, 10)

trust_accuracy_proposed = [85, 88, 90, 92, 94, 95, 96, 97, 97, 98]
trust_accuracy_existing = [70, 72, 74, 76, 78, 79, 80, 81, 82, 83]

# Now plot
plt.figure()
plt.plot(devices, trust_accuracy_proposed, marker='o', label='Proposed Model')
plt.plot(devices, trust_accuracy_existing, marker='s', label='Existing Model')

plt.xlabel("Number of Devices")
plt.ylabel("Trust Accuracy (%)")
plt.title("Scalability Analysis")
plt.legend()
plt.grid()

plt.show()
```

```python
security = [50, 40]
performance = [30, 35]
efficiency = [20, 25]

labels = ['Proposed', 'Existing']

plt.figure()
plt.bar(labels, security, label='Security')
plt.bar(labels, performance, bottom=security, label='Performance')
bottom2 = [i+j for i,j in zip(security, performance)]
plt.bar(labels, efficiency, bottom=bottom2, label='Efficiency')

plt.ylabel('Contribution (%)')
plt.title('System Contribution Analysis')
plt.legend()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

labels = ['Accuracy', 'Latency', 'Detection', 'Scalability', 'Security']
proposed = [96, 85, 95, 90, 94]
existing = [80, 60, 77, 70, 75]

angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False).tolist()

proposed += proposed[:1]
existing += existing[:1]
angles += angles[:1]

plt.figure()
ax = plt.subplot(111, polar=True)

ax.plot(angles, proposed, label='Proposed')
ax.plot(angles, existing, label='Existing')

ax.fill(angles, proposed, alpha=0.1)
ax.fill(angles, existing, alpha=0.1)

ax.set_thetagrids(np.degrees(angles[:-1]), labels)
plt.title("Multi-Metric Comparison")
plt.legend()

plt.show()
```
